# Lyrics Cleaning
This notebook cleans the Eminem lyrics dataset by removing metadata headers (e.g., [Verse 1], [Eminem:]).

```python
import re
import os

input_path = '../data/all_eminem.txt'
output_path = '../data/cleaned_eminem.txt'

if not os.path.exists(input_path):
    print(f"Input file not found at {input_path}")
else:
    with open(input_path, 'r', encoding='utf-8') as f:
        lyrics = f.read()

    # Remove metadata headers in brackets [ ... ]
    # This regex matches anything between brackets, including the brackets themselves.
    cleaned_lyrics = re.sub(r'\[.*?\]', '', lyrics)

    # Optional: Remove extra whitespace/empty lines created by removals
    cleaned_lyrics = re.sub(r'\n\s*\n+', '\n\n', cleaned_lyrics).strip()

    with open(output_path, 'w', encoding='utf-8') as f:
        f.write(cleaned_lyrics)

    print(f"Cleaned lyrics saved to {output_path}")
```

```python
# Display first few lines of cleaned lyrics
if os.path.exists(output_path):
    with open(output_path, 'r', encoding='utf-8') as f:
        for _ in range(20):
            print(f.readline(), end='')
```
